# 03 - Instance Format and Submission Schema Checks

**Competition:** Solar Filament Segmentation Challenge 2026  
**Stage:** Structural verification only; no model predictions or CSV submission are generated.

```python
from pathlib import Path
import sys
import numpy as np
from filament_seg.utils import load_config, get_device
from filament_seg.submission import encode_coco_counts, decode_coco_counts, validate_submission_rows

cwd = Path.cwd().resolve()
project_root = cwd if (cwd / "src" / "filament_seg" / "submission.py").is_file() else cwd.parent
assert (project_root / "src" / "filament_seg" / "submission.py").is_file(), cwd
assert Path(sys.executable).resolve() == (project_root / ".conda" / "python.exe").resolve()
config = load_config(project_root / "configs" / "baseline.yaml")
device = get_device(config["training"].get("device"))
test_dir = project_root / "data" / "filament-segmentation-2026" / "MAGFiLO_1.0_Kaggle_2026" / "test" / "test_images"
test_images = sorted(test_dir.glob("*.jpeg"))
assert len(test_images) == 180

# Synthetic structural example only: this is not a model prediction or Kaggle submission.
mask = np.zeros((2048, 2048), dtype=np.uint8)
mask[100:105, 200:210] = 1
counts = encode_coco_counts(mask)
assert np.array_equal(decode_coco_counts(counts, mask.shape), mask)
row = {"filament_id": f"{test_images[0].stem}_1", "segmentation_rle": counts}
assert validate_submission_rows([row], [path.name for path in test_images]) == 1
print(f"Python: {sys.executable}; device: {device}")
print("Compressed COCO RLE and submission-row structure: PASS (synthetic mask only)")
```
